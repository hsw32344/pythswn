# python数值运算

## 基础数值类型及相关规则
- `int` 整数，没有小数部分
- `float` 浮点数，可以表示小数
- 当 `int` 和 `float` 混合进行普通算术运算时，结果通常提升为 `float`，这样才能保留小数部分
- `float`内部保存的是最接近目标值的近似数，运算后误差可能显示出来，使用 `round()` 可以控制展示或业务舍入结果

### `round()`
- 不能修复浮点数的底层近似,只是把当前近似值舍入，底层仍然是二进制浮点数
- `round()` 在精确平局时采用“取最近值，恰好居中则取偶数
    `round(2.5)  # 2`
    `round(3.5)  # 4`

### 真实场景的近似需求
|场景|解决方法|
|----|----|
|只控制显示的小数位|格式化字符串|
|判断两个计算结果是否足够接近|`math.isclose()`|
|金额、税率等精确十进制业务|`Decimal`，并明确舍入规则|
|固定只有“分”的金额|用整数保存最小货币单位|

## 七个基本运算符
|运算符|含义|
|----|----|
|`+`|加法|
|`-`|减法|
|`*`|乘法|
|`/`|真除法，即使两个操作数都是整数，`/` 也返回 `float`|
|`//`|地板除法，取不大于真实商的最大整数|
|`%`|取余|
|`**`|乘方|

## 数值类型转换
- 转成整数`int()`
    - 朝零截断小数部分
    - 但字符串不是合法的整数字面量时无法转换，必要时先转成浮点数，但是要承担朝零截断的后果
- 转成浮点数`float()`
- 转换函数是在根据输入构造新的目标类型对象，不是修改原对象

## 常用数值内建函数
|函数|机制|
|----|----|
|`round(x, n)`|把数值舍入到指定小数位|
|`abs(x)`|取得绝对值|
|`pow(a, b)`|计算 `a ** b`|
|`pow(a, b, m)`|计算乘方后的模，效果对应 `(a ** b) % m`，但直接执行模幂计算|

### 为什么`int('7.8')` 不自动向零截断，而是报错
- Python 先看“当前对象是什么类型”，再调用该类型对应的舍入、转换或运算协议
- Python 不自动替你串联，是因为这会偷偷丢失信息
- 若明确要求“允许小数字符串，然后截断”，必须显式写出两段转换

### 向下取整、取最近值、向零截断 回顾总结

| 机制 | 控制规则 | `2.7` | `-2.7` | Python 写法 |
|---|---|---:|---:|---|
| 向下取整 | 取不大于原数的最大整数 | 2 | -3 | `math.floor()`、`//` |
| 取最近值 | 取距离最近的整数 | 3 | -3 | `round()` |
| 向零截断 | 丢弃小数部分，向 0 靠近 | 2 | -2 | `int()`、`math.trunc()` |

- “向下”是沿数轴向负无穷方向
- `round()` 是“取最近值，平局取偶数”，不是所有场景中的传统“四舍五入”
- 正数情况下，向下取整和向零截断结果相同；负数情况下才显出本质区别


## 增强赋值
- 增强赋值把“读取旧值、执行运算、更新变量”合并到一条语句中
    - 读取名字当前绑定的对象
    - 执行原来的数学运算
    - 产生结果对象
    - 把名字更新为结果对象
- 增强赋值不会绕过类型限制,字符串仍然只支持它本身定义的运算，例如拼接和重复

|增强赋值|普通写法|
|----|----|
|`x += y`|`x = x + y`|
|`x -= y`|`x = x - y`|
|`x *= y`|`x = x * y`|
|`x /= y`|`x = x / y`,仍然使用 `/` 的规则，所以结果是 `float`|
|`x //= y`|`x = x // y`|
|`x %= y`|`x = x % y`|
|`x **= y`|`x = x ** y`|
