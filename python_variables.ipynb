# python变量
## 定义
- 变量是命名空间字典里的一个条目，由变量名（键）和当前绑定的引用（值）共同组成
    - 纯粹物理层面只有“键”和“值”，不存在叫“变量”的独立实体
- 赋值的本质是绑定，而非创建
    - 它将左侧的变量名绑定到右侧表达式返回的对象引用上
    - 只有当右侧表达式确实生成了新实例时，才会创建对象；若执行 b = a，则不创建新对象，仅将 b 这个键指向 a 已有的对象
    - 绑定完成后，引用指向堆（Heap）中的对象
- 对象的类型和值都存储在对象自身的元数据（PyObject 头部）里，变量名本身没有任何类型信息，因此 Python 是“对象有类型，变量无类型”
- 解释器在运行时通过对象头动态识别类型，故无需声明类型关键字

## 命名规则
- 只能含字母、数字、下划线且不能以数字开头
- 大小写敏感
- 不能用保留字
- 命名约定：小写+下划线

## print()函数
- 完整签名是 `print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)`
- 用逗号分隔时，自动加空格（sep默认值）
- 当参数是字符串时，使用`""`把字面量标记为字符串字面量，消除歧义（避免解释器将其理解为变量）

```python
print("hello world")
print ("A","B","C")
print(1)
```

- 本质上是 `print(sys.stdout.write(sep.join(str(o) for o in objects) + end))`
- 位置参数被 `*objects` 收集打包成元组 `('A', 'B', 'C')`
- 对每个对象调用 `str()` 得到字符串表示
- 用 `sep`（默认一个空格）连接它们，末尾追加 `end`（默认换行），写入 `file`（默认 `sys.stdout`）
