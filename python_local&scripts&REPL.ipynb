# Python本地运行
## 终端命令名解析
- 命令名是你在终端里输入、用来指定“要运行什么”的文字
- 终端把命令名解析为某个可执行文件
    - 可执行文件是 Windows 可以直接启动为进程的程序文件
    - 通常以 .exe 结尾
- 如果只输入`python`，且系统中可能存在多个名为 python 的候选目标时， PowerShell 会根据命令解析优先级和 PATH 顺序选择其中一个执行：
    - 某个 python.exe；
    - Windows 应用商店的 Python 别名
    - PATH 中另一个 Python
    - 同名的 PowerShell 别名或函数
- 因此最好使用完整的路径来指定解释器
    - `& 'C:\project\.venv\Scripts\python.exe'`
    - 引号把含空格或特殊字符的路径保存为一个完整字符串
    - 由于引号中的路径只是一个字符串，PowerShell使用调用运算符 `&` 将它作为命令执行

## Python运行模式
同一个 python.exe，因为收到的参数不同，会进入不同模式

| 命令形状 | Python的行为 |
|---|---|
| `python.exe` | 进入 REPL |
| `python.exe main.py` | 执行文件中的代码 |
| `python.exe -c "代码"` | 执行参数中的代码字符串 |

### REPL模式
- 只启动解释器、不传入脚本文件时，会进入交互式 Shell
- 看到`>>>`，说明解释器正在等待输入
- 输入后，会执行REPL循环
    - Read：读取
    - Evaluate：解析并计算表达式
    - Print：显示结果（其他两个都不会主动显示结果）
    - Loop：回到 >>>，等待下一条输入
- REPL 进程在当前会话中保持状态，执行exit()退出后，内存中的变量消失

### 执行脚本文件代码
- Python先读取和编译整个脚本;编译成功后，再从第一条语句开始按控制流执行
- 最好使用完整的路径来指定脚本文件
- 文件执行结束后退出

### 执行参数中的代码
- 当使用`-c`参数传入代码时，进入该模式
- 把 `-c` 后面的字符串作为代码编译并执行
- 字符串执行结束后退出
