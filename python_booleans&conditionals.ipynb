# 布尔值和条件语句
## 布尔值的来源（简单判断）
- 比较两个对象，直接产生布尔值
    比较运算符： ==、!=、>、<、>=、<=
- 对一个普通对象做真值测试，把普通对象解释为真假
    if 后面不一定要直接写一个布尔值，可以放一个普通对象，让Python来判断

### 对象的真值性质
- truthy/falsy值描述一个对象进入布尔语境时，会被 Python 判断为真还是假
- 当一个对象进入布尔语境， Python 会进行真值测试，当结果为True时其真值性质就是truthy，反之就是falsy
- truthy/falsy 不是对象中保存的属性，是普通对象经过真值测试后才得到的
- 布尔值本身也有真值性质，False是falsy对象，True是truthy对象

### 常见的falsy对象（及truthy对象）
| 类型 | falsy 条件 |
|---|---|
| 数值 | 数值等于零，`0`、`0.0` |
| 容器 | 容器为空，`''`、`[]`、`{}`、`set()` |
| 特殊对象 | `False`、`None` |

- 底层规则
    - 有 `__bool__`() → 使用其返回值
    - 没有 `__bool__`() 但有 `__len__`() → 长度 0 为 falsy
    - 两者都没有 → 默认 truthy

## 布尔值的组合（组合判断）
### 逻辑运算符的机制
- and，从左到右检查左侧对象的真值

    左侧是 falsy，立即返回左侧对象；
    左侧是 truthy，返回右侧对象（如果仍含and，则重复操作）

- or，从左到右检查所有对象的真值

    遇到第一个 truthy 对象，立即返回它；
    如果全部为 falsy，返回最后一个对象

- not，反转结果，返回布尔值

    取得对象，进行真值测试；
    反转结果；
    返回布尔值

### 逻辑运算符的说明
- 当原始操作数是比较运算符串联的组合时，比较好理解（原始操作数全为布尔值）
- and/or 通常返回原始操作数，not 一定返回布尔值
- and/or 都是短路判断，and返回第一个falsy对象，or返回第一个truthy对象
- 通过dataclass创建一个业务检查类，可以把原始值和业务判断封装成一个具有业务真值的操作数，再让 and/or 返回这个完整操作数

```python
from dataclasses import dataclass
from typing import Any


@dataclass
class Check:
    name: str
    value: Any
    ok: bool

    def __bool__(self): # 定义自身真值规则
        return self.ok

student = {
    'name':"Albert",
    'age':"1",
    'job':'teacher'
}

name_check = Check(
    name = 'name_positive',
    value = student['name'],
    ok = bool(student['name']) # 这里可以替换为具体的业务逻辑
)

age_check = Check(
    name = 'age_positive',
    value=student['age'],
    ok = int(student['age']) > 21
)

job_check = Check(
    name = 'job_positive',
    value=student['job'],
    ok = bool(student['job']) # 这里可以替换为具体的业务逻辑
)

result = name_check and age_check and job_check
if result:
    print("ALL GOOD")
else:
    print(f"The first failure is at {result.name}")
    print(f"Its raw value is {result.value}")


```

## 条件语句（控制）
- 条件语句中的控制关键字负责：

    对条件表达式的最终结果进行真值测试；
    然后选择代码路径

具体分工
| 关键字 | 作用 |
|---|---|
| `if` | 检查第一个条件，决定是否进入if代码块 |
| `elif` | 前面的条件为假时，继续检查，决定是否进入elif代码块 |
| `else` | 前面的条件全部为假时，进入else代码块 |

- if 和 elif 必须有条件表达式
- else 不需要条件表达式，而且不能写
