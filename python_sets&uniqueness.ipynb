# 集合和集合方法

## 集合的特征
- 集合元素无序
    - 无序指不能依赖它的排列顺序，不是每次打印一定随机打乱
- 集合元素唯一
- 集合元素必须可哈希
    - 可哈希
        - 可哈希 ≈自身不可变+内部嵌套的相关对象也必须不可变
        - 可哈希即能够提供一个稳定的哈希值
    - 集合和字典底层都是哈希表，需要稳定和哈希值来稳定定位
    - 集合可以视作一个只有键的字典

## 集合的创建
- 直接用集合字面量创建，但是不能用于创建空集合
- `set(iterable)` 把可迭代对象转为集合
    - 空集合用 `set()` 创建， `{}` 建立的是空字典
    - 用于去重；iterable转成集合后，重复值只剩一份
- 集合推导式
    - `{expression for item in iterable if condition}`
    - 不同于`{key:value for item in iterable if condition}`

```python
s = set(range(5))
print(s)
s =set()
print(s) # 显示为 set() ，表示这是一个空集合，不是空字典
```

```python
cities = ["常州", "苏州", "常州"]
unique_cities = set(cities)

print(len(cities))         # 3
print(len(unique_cities))  # 2
```

## 集合的增删
- 集合可以增删成员，重复添加已有成员不会增加数量

### 增加元素
- `s.add(x)` 把参数 x 当作一个完整成员
- `s.update(iterable)` 遍历参数，把读取到的元素分别作为成员加入
- 两者都修改 `s` 指向的集合，返回 None
- 重复元素不会多占一份,重复添加已有成员不会增加数量

```python
fields = {"id"}

fields.add("amount")
print(fields)  # 'amount', 'id'

fields.update("amount")
print(fields)  # 'm', 'id', 'amount', 'o', 'n', 'a', 't', 'u'

fields = fields.add("city")  # 不要这样接回原变量
print(fields) # None
```

### 删除元素
- `s.remove(item)` 和 `s.discard(item)` 接收一个要删除的成员，并修改原集合
- 目标不存在时，remove 把它视为失败，discard 接受“已经不存在”这个状态，不报错

- 先查找目标找到则删除
    - 正常返回 None
    - 未找到时
        - remove 抛出 KeyError
        - discard 保持集合不变并返回 None


- `s.pop()` 集合的pop方法不支持指定成员（不传参数），随机删一个并返回该成员，不是用于随机抽样的接口
    - `dict.pop(key[,default])` 字典的pop方法必须指定成员，如果不存在会返回default参数设定的值（如果default也没设置，就报错）
    - `ls.pop()`列表的pop方法可以指定一个索引位置，如果不指定就默认弹出最后一个元素`ls[-1]`
- `s.clear()` 删除所有成员，不传参数，返回 None

## 集合的成员判断
- 返回布尔值

### 成员与集合
- x in s 判断一个成员是否存在

### 集合间包含、真包含
- A <= B 判断 A 的每个成员是否都在 B 中，返回布尔值，不修改集合
- A >= B 反向判断包含关系
- 同理 A > B, A < B 判断真包含关系
- 注意：集合比较不是比较成员数量（长度比较）

### 判断有无交集
- A.isdisjoint(B) 判断是否完全没有共同成员
    - disjoint = 不相交的、没有交集的
    - A.isdisjoint(B) -> 两个集合是否不交？
        - 完全没有共同元素 -> True
        - 只要有一个共同元素 -> False


## 集合运算
- 返回一个新的集合

| 运算 | 含义 | 保留规则 | 方法形式 |
|--------|------|------|----------|
| `A \| B` | 并集 | 任意一方存在 | `A.union(B)` |
| `A & B` | 交集 | 双方都存在 | `A.intersection(B)` |
| `A - B` | 差集 | 只保留左侧独有 | `A.difference(B)` |
| `A ^ B` | 对称差 | 只在其中一方存在 | `A.symmetric_difference(B)` |

- 集合二元运算都会返回新集合，不修改原集合
- `A - B` 有方向，表示从 A 中找 B 没有的成员
    - 字段校验常用：
        - `required - actual`：需要，实际没有 -> 缺失字段
        - `actual - required`：实际有，不需要 -> 额外字段
- `A ^ B` 只能找出双方不同的成员，不能区分“缺失”还是“额外”
- 记忆：`|` 汇总，`&` 共同，`-` 左侧独有，`^` 双方不同。
