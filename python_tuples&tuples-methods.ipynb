# 元组和元组方法

## 元组特征
- 有序
- 元素可异构，约定用于存放异构数据，即表示一条固定记录（Record）的不同属性
- 结构不可变，即创建后不能增删或者变换顺序
- 使用0基索引

## 元组创建
- 元组字面量的核心是逗号
- 圆括号通常用于明确边界，但单独的圆括号也可能只是普通表达式分组
- 此外还可以通过`tuple(iterable)`内置函数，把可迭代对象转化为元组

| 写法 | 分类 |
|---|---|
| `(a, b, c)` | 元组字面量语法 |
| `(a,)` | 单元素元组语法 |
| `tuple(iterable)` | 内置类型构造 |

```python
rule_ids = ("DQ-001",)
not_a_tuple = ("DQ-001")

print(type(rule_ids))
print(type(not_a_tuple))
```

```python
source_columns = ["customer_id", "phone", "created_at"]

fixed_columns = tuple(source_columns)

print(fixed_columns)
```

## 元组不可变性
- 元组创建后，其位置数量以及每个位置引用哪个对象不能被重新安排
- Python不允许对元组执行元素赋值、元素删除、元素增加、元素变序等
- 不可变约束的是元组自身的引用结构，不会自动冻结内部可变对象
    - “元组不可变”不代表元组引用的所有内部对象都会递归冻结
    - 允许修改元组内位置引用的可变数据类型，此时元组里的位置仍然指向原数据对象

```python
rule = ("DQ-001", ["customer_id"])

rule[1].append("order_id")

print(rule)
```

## 元组读取
- 元组和列表一样采用零基索引；正索引从左侧 0 开始，负索引从右侧 -1 开始
- 使用位置读取元组中的一个对象`tuple[index]`
- 使用切片操作从原元组选择一段位置，构造新的元组`tuple[start:stop:step]`，切片操作享受截断机制

## 元组元素查询
- `in` 成员判断
- `tuple_object.count(value)`：从头到尾扫描元组，统计与目标值相等的元素数量
- `tuple_object.index(value, start, stop)`：搜索指定范围，返回首次匹配位置

## 元组解包
- 按位置读取元组元素，并分别绑定到左侧名字
- 普通解包要求变量数量与元素数量相等
- 带星号的变量收集剩余元素，并生成列表

```python
rule = ("DQ-001", "customer_id", "NOT_NULL")

rule_id, field_name, check_type = rule # 普通解包

print(rule_id)
print(field_name)
print(check_type)
```

```python
rule = (
    "DQ-001",
    "customer",
    "customer_id",
    "NOT_NULL",
    "HIGH",
)

rule_id, *rule_details, severity = rule # 星号解包

print(rule_id)
print(rule_details)
print(severity)
```

## 元组变序重构(输出列表)
- 读取元组元素，根据元素本身或 key 结果排序，返回新的列表
- `sorted()` 是内置函数，不是元组方法
- `sorted()` 返回列表，而不是元组。如果结果仍需固定为元组，可以显式重新构造

```python
rules = (
    ("DQ-003", 2),
    ("DQ-001", 3),
    ("DQ-002", 1),
)
# sorted()进行排序时，会依次遍历该对象中的每一个元素
# 并对每个元素执行 key 参数指定的函数
ordered_rules = tuple(sorted( # 显式转化为元组
    rules,
    key=lambda rule: rule[1],
    reverse=True,
))

print(ordered_rules)
```
