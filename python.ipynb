# python
## 定位
- general-purpose编程语言，以简洁易用著称
### 通用性
- 具备图灵完备的计算能力
- 覆盖文件、网络、进程、数据库、数据结构的标准库（"batteries included"，自带电池）和跨平台解释器
### 简洁易用性
- 用缩进表达代码块，语法可读，但排版受限
- 动态类型，写起来快，类型错误推迟到运行到那一行才暴露
- 解释执行，执行速度慢于 C/Java，换来快速迭代和跨平台可移植
- 自动内存管理，出错率低，但是吞吐不如手工管理
### 数据科学中的生态分工
- Python负责编排，扮演胶水代码
- 数据交给C内核实现，更快

## 衍生技术栈（多生态性）
- 数据科学/ML：Pandas、NumPy；TensorFlow、Scikit-learn
- Web开发：	Django、FastAPI、Flask，做数据接口/API 服务
- DevOps：数据管道部署和运维
- 自动化/爬虫：Selenium、BeautifulSoup，数据采集工作
- 软件测试：pytest，做数据校验
