# 字符串

## 定义与性质
- 字符串是由单引号或双引号包围的一系列字符
- 字符串是 Python 中的不可变数据类型，创建对象后就不能修改原始对象本身

## 字符串方法
- 方法是属于特定对象/类的函数，写法 `对象.方法(参数)`，后面函数和OOP时讲底层机制
- 字符串方法其实是绑定在 str 类上的函数
- 字符串方法不修改原串，全部返回新字符串或布尔值等

### 清洗与标准化
|方法|用法|
|----|----|
|strip()|清洗数据，去除首尾空格、换行、制表符|
|lower()/upper()|统一大小写，便于匹配和去重|
|replace()|替换脏字符，删除指定字符、符号|
|split()|把复合文本拆为要求的词标签等|
|join()|组合多个字符串为一段文本|

```python
raw_name ='    Jack Smith ' # 左边是制表符，右边是空格
clean_name = raw_name.lower().strip().strip(    ).replace(' ','_')
# 对于连续的点号调用，Python 先执行最内层/最左侧，再依次向右执行
print(clean_name)
```

```python
raw_date ='26/8/18'
raw_date_tokens = raw_date.split('/')
print (f'split方法输出一个拆分碎片组成的列表：{raw_date_tokens},{type(raw_date_tokens)}')
standardized_date = '-'.join(raw_date_tokens)
print(f'join方法按调用字符串作为粘合剂拼接可迭代对象：{standardized_date},{type(standardized_date)}')
```

### 匹配和过滤
|方法|用法|
|----|----|
|startswith()|判断前缀是否匹配，例如编码开头或者路径|
|endswith()|判断后缀是否匹配，例如文件扩展名|
|find()|返回字串第一次出现的位置，找不到返回-1|
|count()|统计某个字串出现的次数|
- find方法通常没有in运算符快捷，如果需要知道位置，才用 find()
- 关于子串的检索、匹配和统计可以参考jieba分词和正则

```python
path = 'C:/correctpath/A.xlsx'
path_valid = path.startswith('C:/correctpath')
print(f'path is valid: {path_valid}')
```

```python
filename = 'A.xlsx'
file_valid = filename.endswith('.xlsx')
print(f'file is valid: {file_valid}')
```

### 显示格式和输出
|方法|用途|
|----|----|
|upper()|统一状态码、识别码、分类标签|
|title()|规范化人名或标题名|
|capitalize()|首字母大写，规范化简单文句|

```python
raw_code = "Paid-12345-26-8-23"
raw_code_tokens = raw_code.split('-')
if len(raw_code_tokens) == 5:
    standardized_code_status = raw_code_tokens[0].upper()
    standardized_code_number = raw_code_tokens[1]
    raw_code_date = raw_code_tokens[-3:]
    standardized_code_date = "".join(raw_code_date)
    standardized_code = f"{standardized_code_status}-{standardized_code_number}-{standardized_code_date}"
else:
    print(f"原编码格式非法")

print(standardized_code)

```

## 索引与切片
- 利用python序列定位的相关坐标体系，可以对字符串进行取值和切片
- 取值操作`s[index]`
- 切片操作`s[start_index:stop_index:step]`，包含start，但是不包含stop
- 可以省略 start 和 stop 索引，start 默认取行进方向的第一个元素，stop 默认取行进方向尽头的外边界
- step 默认取1，代表正向连续取值（每次偏移1）
- 常用：
    1. `s[:n]`       # 前 n 个字符
    2. `s[-n:]`     # 后 n 个字符
    3. `s[:-n]`      # 删除后 n 个字符
    4. `s[::-1]`     # 反转字符串

## 串接
- 使用加号（+）操作符将多个字符串串接在一起
- 只能拼字符串，且Python 不做隐式类型转换，必须显式 str()转换
- str() 返回对象的字符串表示，不修改原对象
- 用 * 操作符将字符串乘以一个整数来重复它
- 可以使用扩充赋值（+=）操作符来进行串接，一步实现扩充和赋值

## f字符串实现字符插值
- F-strings 以 f（小写或大写）开头，紧跟引号，允许在由花括弧`{}`中嵌入变量或表达式，自动转字符串、无需 str()
- f-string 在编译期被解析为格式化的字节码，花括号内的表达式在运行时求值，结果通过 __format__ 协议转成文本

## 其他知识点
- 单/双引号等价（Python 不区分，区别于 C/Java）
- 含引号时用另一种引号包裹，或反斜杠转义
- 三引号 """..."""/'''...''' 定义多行字符串
- len(s) 求长度
- str() 返回对象的字符串表示，不修改原对象
- in 做子串成员检测（返回布尔）
