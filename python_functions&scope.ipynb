# 函数和作用域

## 输入函数 `input()`
`input()`用于接收用户输入
- 暂停程序，等待用户输入
- 用户输入进入程序后，以字符串形式返回

## 自定义函数
- 自定义函数使用 def，函数定义包含函数名、参数和函数体
- 定义函数后，需要通过函数名加括号，传入实参调用
    - 参数是函数定义中的局部名字，实参是调用时提供的实际对象
    - 调用发生时，实参按照调用规则绑定到参数，函数体通过参数访问传入对象
- 函数可以使用 return 返回结果；如果函数没有显式执行 return，默认返回 None

```python
amount = input('Enter amount: ')
print(amount,type(amount))
```

```python
def show_message(message):
    print(message)

result = show_message('Hello') # 调用时执行函数体，print("Hello")
print(result) # 无显式return，返回 None
```

## 作用域

### 局部作用域和全局作用域
- 作用域是查找变量的指导书，规则是LEGB，局部作用域、外围作用域、全局作用域、内置作用域
- 区别于命名空间，作用域静态的，是在编译时（写代码时）就确定的语法结构
- 函数参数以及函数体内部创建的变量属于局部作用域，只能在对应函数内部使用
- 函数外定义的名字属于全局作用域，函数内部可以读取可见的全局变量

### 外围作用域
- 指的是“嵌套在外层的那个函数的局部命名空间”
- 只要代码缩进是嵌套的，Enclosing 作用域就一定存在，无论内层函数用不用它
