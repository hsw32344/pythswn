# 字面量
## 定义
- 字面量是直接写在源代码中，用固定语法表示一个值的写法
- Python解析代码时，先根据字面量的写法识别它表示的值和类型；运行到该表达式时，取得对应的对象
    - `42`表示整数值，得到`int`对象
    - `"42"`表示由两个字符组成的文本，得到`str`对象
    - `True`和`None`分别表示布尔真值和没有值对应的单例对象
- 字面量本身不负责给对象命名，赋值语句才会把左侧名字绑定到右侧字面量求值得到的对象
- 字面量不需要先去命名空间查找；变量名必须先在当前作用域对应的命名空间中解析

## 字面量、变量名和构造调用
| 写法 | 分类 | 执行时发生的事 |
|---|---|---|
| `42` | 整数字面量 | 取得值为42的`int`对象 |
| `"42"` | 字符串字面量 | 取得内容为42的`str`对象 |
| `amount` | 变量名 | 去命名空间查找`amount`当前绑定的对象 |
| `int("42")` | 函数调用表达式 | 先取得`int`和字符串对象，再调用`int()`构造新的整数结果 |
| `[1, 2]` | 列表显示式（常称列表字面量） | 先取得元素对象，再建立新的列表对象 |
| `set()` | 内置类型调用 | 调用`set`建立空集合，不是字面量 |

```python
amount = 42 # 42是整数字面量，amount是接收绑定的变量名
copied_amount = amount # amount是名字，需要从命名空间读取已有对象
text_amount = "42" # 引号使42成为字符串内容
converted_amount = int(text_amount) # 函数调用根据字符串构造整数对象

print(amount,type(amount))
print(text_amount,type(text_amount))
print(converted_amount,type(converted_amount))
print(amount is copied_amount)
```

## 常见字面量和显示式
| 值的种类 | 写法示例 | 结果类型 |
|---|---|---|
| 整数 | `10`、`3`、`0xFF` | `int` |
| 浮点数 | `3.14`、`1e3` | `float` |
| 虚数 | `2j` | `complex` |
| 字符串 | `"hello"`、`'hello'`、`f"{amount}"` | `str` |
| 字节串 | `b"ABC"` | `bytes` |
| 布尔值 | `True`、`False` | `bool` |
| 空值 | `None` | `NoneType` |
| 列表显示式 | `[1, 2, 3]` | `list` |
| 元组显示式 | `(1, 2, 3)`、`(1,)` | `tuple` |
| 集合显示式 | `{1, 2, 3}` | `set` |
| 字典显示式 | `{"name": "Albert"}` | `dict` |

- 严格按照Python语言参考，数字、字符串和字节串使用literal（字面量）这个术语；列表、元组、集合和字典的语法称为display（显示式）
- 日常交流中经常把这些显示式统称为容器字面量，但需要知道它们会先计算内部表达式，再建立容器对象

```python
number_literal = 10
float_literal = 3.14
string_literal = "hello"
boolean_literal = True
none_literal = None
list_display = [1, 2, 3]
tuple_display = (1, 2, 3)
set_display = {1, 2, 3}
dictionary_display = {"name": "Albert"}

for value in (number_literal,float_literal,string_literal,boolean_literal,none_literal,list_display,tuple_display,set_display,dictionary_display):
    print(repr(value),type(value))
```

## 容器显示式的执行顺序
- 容器的外层结构直接写在代码里，但内部位置可以放字面量、变量名、运算和函数调用等表达式
- Python运行到容器显示式时：
    1. 按表达式的求值顺序取得内部元素或键值对应的对象
    2. 建立新的容器对象
    3. 让容器中的位置保存对这些对象的引用
- 因此`[amount, 1 + 2]`不是一整块完全固定的数据：`amount`需要名字查找，`1 + 2`需要执行加法，之后才建立列表
- 创建容器后修改容器，不会反过来修改源代码中的字面量

```python
amount = 10
values = [amount,1 + 2,"fixed"]

amount = 20 # 重新绑定amount，不会改写列表中已经保存的整数对象引用
values.append(30) # 修改运行时的列表对象，不会改写源代码

print(amount)
print(values)
```

## 容易混淆的边界
- `{}`建立的是空字典，不是空集合；空集合只能写成`set()`
- 元组的核心语法是逗号：`(1,)`是单元素元组，`(1)`只是加括号的整数表达式
- `-3`求值后是负整数，但语法上是一元负号作用于整数`3`；`1+2j`同样是加法表达式，不是一个完整的复数字面量
- 字符串中的内容不会再被当作Python代码执行：`"1 + 2"`是五个字符组成的字符串，不会自动得到整数3
- 同样的不可变字面值可能被解释器复用对象，这属于实现优化；不能依靠`is`判断两个数字或字符串的值是否相等，值比较应该使用`==`
- 字面量只说明代码怎样写出值，不代表得到的对象一定可变或不可变；对象能否修改由其类型决定

```python
empty_mapping = {}
empty_unique_values = set()
one_item_tuple = (1,)
grouped_integer = (1)
code_text = "1 + 2"

print(type(empty_mapping))
print(type(empty_unique_values))
print(type(one_item_tuple))
print(type(grouped_integer))
print(code_text,type(code_text))
```
