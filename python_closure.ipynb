# 闭包
## 闭包（Closure）定义
- 闭包是一个函数对象，附带了一个叫 `__closure__` 的隐藏数据包
- 当仍然存活的内部函数对象所需要的变量在外部函数时(在自身的局部作用域找不到)，
- Python会在运行时把这个变量保存到堆中的某个容器中，好让它在外部函数的局部命名空间销毁后仍存在，
- 在命名空间的视角，这个变量属于外部函数的局部命名空间
- 在作用域的视角，由于内部函数在Local中找不到，又不属于Global（顶格写的那部分），因此定位为Enclosing外围作用域，表示围住Local的域
- 只要代码缩进是嵌套的，Enclosing 作用域就一定存在，无论内层函数用不用它


|LEGB 层级|对应的命名空间（实际字典）|
|----|----|
|L（Local）|局部命名空间（当前函数专属）|
|E（Enclosing）|外层函数的局部命名空间（被闭包捕获后残存的那部分）|
|G（Global）|全局命名空间（整个文件模块专属）|
|B（Built-in）|内置命名空间（Python 解释器自带）|

```python
def outer(x):
    def inner(y):
        return x+y
    return inner
add_x = outer(x=5)
print(add_x(1))
```

```python
def bad_idea():
    to_eat = []
    foods = ['apple','banana','carrot']
    for food in foods:
        to_eat.append(lambda:food) # 内层函数不接受参数，返回food（这个food是外围作用域的变量）
    return to_eat

to_eat = bad_idea() 
# to_eat里的三个匿名函数用到了enclosing中的food变量
# 随着外层函数的销毁，保留到各自的__closure__数据包中
print(to_eat[0]()) # 经过循环过后，food停在了最后一个carrot上，
print(to_eat[1]()) # 由于内层函数逻辑要求去找外围作用域中的变量，
print(to_eat[2]()) # 因此全为carrot
```

```python

def good_idea():
    to_eat = []
    foods = ['apple','banana','carrot']
    for food in foods:
        to_eat.append(lambda food_snapshot=food:food_snapshot) # 内层函数使用food作为默认值，返回这个默认值
        # 相当于分别传入了'apple','banana','carrot'作为实参，保留到各自的__defaults__数据包中
    return to_eat

to_eat = good_idea()

print(to_eat[0]()) # 由于此时并没有引用到enclosing的变量，所以这些函数对象不算闭包
print(to_eat[1]())
print(to_eat[2]())
```
