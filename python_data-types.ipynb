# 数据类型
## 静态类型（C#/Java/C++）
- 编译期就检查类型，编译不通过不让运行
## 动态类型（python）
- 类型属于对象，不属于变量
- 类型由赋给的值在运行时才决定，无需声明
- 类型检查后置到运行期，逐行运行，直到无法继续后才停下抛出错误

## 常见数据类型和底层本质
|类型|`type()`|底层本质|
|----|----|----|
|`int`|`<class 'int'>`|任意精度整数（大整数不溢出），C层 `PyLongObject`|
|`float`|`<class 'float'>`|IEEE754双精度|
|`str`|`<class 'str'>`|不可变Unicode字符序列|
|`bool`|`<class 'bool'>`|int的子类，二值（`True == 1`，`False == 0`）|
|`set`|`<class 'set'>`|无序、不重复，打印顺序不保证，底层哈希表|
|`dict`|`<class 'dict'>`|键值对，按键访问，保证保持插入序，底层哈希表|
|`tuple`|`<class 'tuple'>`|不可变有序序列，元素可异构，只要是对象引用都能放|
|`range`|`<class 'range'>`|惰性数字序列，只存`start`、`stop`、`step`，迭代到第 i 个元素时，用公式 `start + i * step` 当场算出来|
|`list`|`<class 'list'>`|可变有序序列，元素可异构|
|`None`|`<class 'NoneType'>`|单例，表示“没有值”，判"没有值"要用 `is None` 而不是 `== None`|

## type(x)
- 三态函数，用于检查对象和动态创建新类
- 返回传入参数所属的类对象，无参时失败
- `type(name, bases, dict)`动态创建新类

## isinstance(x,T)
- 沿继承链查找,看x是否是T或者T的子类，检查比 `type()` 更宽松

```python
print(type(True))
print(type(1)) # type() 返回类对象
print(isinstance(True,bool))
print(isinstance(True,int))
print(type(True) == bool)
print(type(True) == int)
```
