# range & enumerate & zip
- 三者都可以向for循环提供元素，但提供的内容不同
- `range`按范围遍历
    - 输入：整数边界规则
    - 输出：一个个整数
- `enumerate`带序号遍历
    - 输入：一个可迭代对象
    - 输出：(计数, 元素)
- `zip`配对、打包遍历
    - 输入：多个可迭代对象
    - 输出：(第1组元素, 第2组元素, ...)

## range()：描述整数序列
### 基础概念
- range 对象保存 start、stop 和 step 三个整数规则，需要某个位置的值时再根据规则计算
    - 计算规则：当前值 = start + 轮次 × step
    - 只要下一项已经到达或越过 stop，序列结束
- `range()` 有三种常用形式
1. `range(stop)` 默认值：start = 0，step = 1
2. `range(start, stop)` 默认值：step = 1
3. `range(start, stop, step)`
- 允许负步长，代表从大到小输出值

```python
def countdown(start_number):
    for tmp_count in range(start_number,0,-1):
        print(tmp_count)
    return None

countdown(5)
```

### 边界/错误
- 步长不能为零
- 三个参数必须都是整数
- 方向与边界不匹配时得到空序列

```python
print(list(range(1, 5, -1)))
print(list(range(5, 1, 1)))
```

### range() 与索引遍历
- 可以range(len(items)) 产生合法索引，再通过索引读取列表元素
- 理由：
    有序序列的合法索引是 0 到 len - 1
    range(len(items))恰好生成 0 到 len - 1 的连续数字
- 但需要计数和元素时，下一节的 enumerate() 通常更直接

```python
names = ["Albert","Beck","David"]
for i in range(len(names)):
    print(f"<{i+1}>{names[i]}")
```

## enumerate()：同时取得计数和元素
- enumerate() 要求一个可迭代对象
- 每次底层输入提供一个元素，它同时产生当前计数和该元素组成的元组
- 常用形式：
1. `enumerate(iterable)` 计数默认从0开始（start=0）
2. `enumerate(iterable, start=0)` 自行选择计数开始起点
- start 是计数起点，不是切片起点，不会跳过任何索引

```python
names = ["Albert","Beck","David"]
for i in enumerate(names,1):
    print(f"<{i[0]}>{i[1]}")
```

```python
for i in enumerate(range(3)):
    print(i) # 迭代过程中返回的是元组
```

## zip()：按位置并行配对（并行迭代）
- zip() 持有多个输入的迭代器
- 每一轮分别从每个输入中取一个元素，然后把这些元素组成元组
- 这些元组再供给下游消费

```python
field_names = [
    "customer_id",
    "phone",
    "created_at",
]

field_values = [
    "C001",
    "123-4567-8910",
    "2026-09-04",
]

records = dict(zip(field_names, field_values))

print(records)
```

- zip() 不提前知道所有输入的长度，它在每一轮真正向各个输入请求下一项
- 某个输入先耗尽时，整个 zip 结束
- Python 3.10及以上可以使用：`strict=True` 参数，在长度不一致时抛出ValueError，避免默认静默截断

```python
columns = ["customer_id", "phone", "created_at"]
values = ["C001", "138****1234"]

pairs = list(zip(columns, values,strict=True))

print(pairs)
```

## enumerate()和zip()实现带索引计数的并行迭代与解包
- zip() 先提供字段和值的配对，enumerate() 再为每一对添加显示编号
- 使用for循环进行元组解包

```python
field_names = [
    "customer_id",
    "phone",
    "created_at",
]

field_values = [
    "C001",
    "123-4567-8910",
    "2026-09-04",
]
# 注意元组解包时拆分的结构是一个单独的循环变量和两个在元组里的循环变量
# 即保证左侧解包结构必须与右侧元素结构匹配
for tmp_No,(tmp_fn,tmp_fv) in enumerate(zip(field_names,field_values,strict=True),start=1):
    print(f"<{tmp_No}> {tmp_fn} = {tmp_fv}")
```
