# 列表与列表方法

## 列表的特征
- 有序序列
- 元素可异构，但通常约定用于存放同构数据
- 可变数据类型
- 使用0基索引

## 创建列表
1. 使用方括号创建
    - `numbers = [10, 20, 30, 40]`
2. 使用`list()`创建列表，`list()`是Python内置函数
    - `list(iterable)`
    - 可迭代对象可以逐个产生元素，`list(iterable)`读取这些元素并创建列表
3. 使用`sorted()`生成排序后的新列表，`sorted(iterable, key=..., reverse=...)`是Python内置函数
    - 读取一个可迭代对象，建立新的排序列表
    - key 默认None，表示直接比较元素本身
        - 字符串走字典序，逐个字符比较其 Unicode 码点
        - 如果列表中保存的是字典，字典本身不能直接比较大小，需要设置key
    - reverse 控制是否反转最终排序方向，默认`reverse=False`表示升序

```python
issues = [
    {"rule_id": "DQ-03", "severity": "medium"},
    {"rule_id": "DQ-01", "severity": "high"},
    {"rule_id": "DQ-02", "severity": "low"},
]
severity_rank = { # 使用字典建立字符串到业务逻辑的映射
    "high":"003", # 填充前导0，使字典序与数值序一致
    "medium":"002",
    "low":"001"
}
new_issues = sorted(issues,key=lambda issue: severity_rank[issue["severity"]],reverse=True)

print(new_issues)
```

## 读取列表（列表的位置模型）
- 列表对象按顺序保存多个位置，每个位置引用一个 Python 对象
- 使用索引读取一个元素，超出有效范围会报错`IndexError`
    - `numbers[索引]`
- 使用切片读取多个元素,构造新列表
    - `列表[开始索引:结束索引:步长]`,可以省略开始、结束位置或步长
    - 切片产生一个新列表，原列表不变
    - `start` 包含，`stop` 不包含
    - 切片端点会根据序列长度和步长方向进行正规化，因此越界端点通常会被截断
- 使用`len()`取得元素数量，`len()`是Python内置函数
    `numbers = [10, 20, 30, 40]`
    `count = len(numbers)`
    `print(count)  # 4`
- 使用`in`判断元素是否存在，`in`是Python语法中的运算符
- 使用`index()`查询位置，`index()`是列表方法，`list.index(x, start, end)`
    - 如果元素不存在，`index()`会报错

### 切片的截断机制
（切片的截断机制由所有内置序列类型共享）
- `stop` > `length`，解释器会将 `stop` 强制重置为 `length`
- `start` < `-length`，解释器会将 `start` 强制重置为 0
-  `start` 和 `stop` 都指向无效位置时，直接返回一个空列表 `[]`

### 访问嵌套列表
- 采用连续的中括号 `[][]` 进行逐层访问

## 修改列表内容

### 使用索引赋值修改某个位置
- `numbers[0]=200`
- 必须使用有效索引
- 修改已有位置不能自动扩展列表

### 向列表添加元素

|方法名|机制|
|----|----|
|`append()`|`列表.append(元素)`，在末尾添加一个元素|
|`insert()`|`列表.insert(索引, 元素)`，在指定位置添加一个元素|
|`extend()`|添加一批元素|

- `append()` 始终把参数作为一个元素添加，如果参数是列表，添加一个列表元素
- `insert()` 安排元素进入索引后，原来索引及后面的元素向右移动
- `extend()` 会逐项读取参数中的元素

### 删除列表元素

- `del`：按位置删除
    - `del items[index]`、`del items[start:stop]`
    - 删除原列表中的指定位置或切片
    - del 是删除语句，不是函数或列表方法；它没有返回值
    
|方法名|机制|
|----|----|
|`remove()`|`列表.remove(元素)`，从左向右搜索目标值，删除第一次相等的元素，返回 `None`|
|`pop()`|`列表.pop(索引)`，删除指定位置，同时把被删除元素返回给调用者|
|`clear()`|`列表.clear()`，原地清空现有列表对象中，并返回 `None`|

- `remove()` 的输入是值，而不是位置；存在重复值时不会全部删除
- `pop()`省略索引时，默认删除并返回最后一项
- `ls = []` 让名字绑定到一个新列表对象；而 `ls.clear()` 则让原列表对象的内容清空

### 列表原地排序

|方法名|机制|
|----|----|
|`sort()`|`items.sort(key=..., reverse=...)`，原地排序，重新排列原列表内部的位置顺序，返回 None|
|`reverse()`|`items.reverse()`，颠倒当前顺序，返回 `None`|

- 需要保留来源顺序时用 `sorted()`；允许改变原列表时才考虑 `sort()`
- `reverse()` 是位置反转；`sort(reverse=True)` 是根据排序规则降序排列

## 普通解包与星号解包
- 读取列表位置中的对象，将它们分别绑定到左侧变量名
- 解包属于赋值语法
    - 普通解包要求左右数量一致
    - 星号解包
        - *name 动态吸收剩余元素并生成列表
        - 赋值语法不允许同时出现2个星号，会导致二义

```python
# 普通解包：结构固定且可知
rule_record = ["DQ-01", "customer_id", "非空"]
rule_id, field_name, requirement = rule_record

print(rule_id)
print(field_name)
print(requirement)

# 星号解包：首尾固定，中间步骤数量灵活可变
governance_flow = [
    "原始数据",
    "字段标准化",
    "敏感字段识别",
    "质量检查",
    "治理报告",
]

source, *processing_steps, output = governance_flow

print(source)
print(processing_steps)
print(output)
```
